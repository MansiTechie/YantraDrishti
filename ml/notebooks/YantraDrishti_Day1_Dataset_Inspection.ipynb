```python
import pandas as pd
import os

path = "/content/CMAPSSData"

# Column names
columns = (
    ["unit_number", "cycle",
     "setting_1", "setting_2", "setting_3"]
    + [f"sensor_{i}" for i in range(1, 22)]
)

# Load training dataset
train_df = pd.read_csv(
    os.path.join(path, "train_FD001.txt"),
    sep=r"\s+",
    header=None,
    names=columns
)

print("Dataset loaded successfully!")
print("Shape:", train_df.shape)

train_df.head()
```

```python
# 1. Check data types and information
print("DATA INFORMATION")
train_df.info()

# 2. Check missing values
print("\nMISSING VALUES")
print(train_df.isnull().sum().sum())

# 3. Check duplicate rows
print("\nDUPLICATE ROWS")
print(train_df.duplicated().sum())

# 4. Statistical summary
print("\nSTATISTICAL SUMMARY")
display(train_df.describe())
```

```python
# Find columns with only one unique value
constant_columns = [
    col for col in train_df.columns
    if train_df[col].nunique() == 1
]

print("Constant columns:")
print(constant_columns)

print("\nNumber of engines:")
print(train_df["unit_number"].nunique())

print("\nOperating cycles per engine:")
print(train_df.groupby("unit_number")["cycle"].max().describe())
```

```python
# Find the last cycle of every engine
max_cycles = train_df.groupby("unit_number")["cycle"].max()

# Calculate RUL for every row
train_df["RUL"] = (
    train_df["unit_number"].map(max_cycles)
    - train_df["cycle"]
)

# Display results
print(train_df[["unit_number", "cycle", "RUL"]].head(10))

print("\nRUL Statistics:")
print(train_df["RUL"].describe())
```

```python
# Load test dataset
test_df = pd.read_csv(
    "/content/CMAPSSData/test_FD001.txt",
    sep=r"\s+",
    header=None,
    names=columns
)

# Load actual RUL labels
test_rul = pd.read_csv(
    "/content/CMAPSSData/RUL_FD001.txt",
    sep=r"\s+",
    header=None,
    names=["RUL"]
)

# Display information
print("TEST DATA SHAPE:", test_df.shape)
print("TEST ENGINES:", test_df["unit_number"].nunique())

print("\nACTUAL RUL SHAPE:", test_rul.shape)
print("\nACTUAL RUL SUMMARY:")
print(test_rul.describe())

print("\nTEST DATA FIRST 5 ROWS:")
display(test_df.head())

print("\nACTUAL RUL FIRST 10 VALUES:")
display(test_rul.head(10))
```

```python
# Get last recorded cycle for each test engine
last_cycles = test_df.groupby("unit_number")["cycle"].max()

# Match each engine with its actual RUL
engine_ids = sorted(test_df["unit_number"].unique())

rul_mapping = dict(zip(
    engine_ids,
    test_rul["RUL"].values
))

# Calculate RUL for every test record
test_df["RUL"] = (
    test_df["unit_number"].map(rul_mapping)
    + test_df["unit_number"].map(last_cycles)
    - test_df["cycle"]
)

print("Test RUL calculated successfully!")

# Verify the final cycle of each engine
last_records = test_df.loc[
    test_df.groupby("unit_number")["cycle"].idxmax()
]

display(last_records[["unit_number", "cycle", "RUL"]].head(10))

print("\nTest RUL statistics:")
print(test_df["RUL"].describe())
```

```python
# Get the final record of each test engine
last_records = test_df.loc[
    test_df.groupby("unit_number")["cycle"].idxmax()
].sort_values("unit_number")

# Compare calculated RUL with original RUL labels
calculated = last_records["RUL"].values
actual = test_rul["RUL"].values

print("All engines matched:", (calculated == actual).all())
print("Total matching engines:", (calculated == actual).sum())
print("Total test engines:", len(actual))
```
